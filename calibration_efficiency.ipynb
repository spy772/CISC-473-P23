# Calibration vs. conformal efficiency (Problem 2)

**Question:** does better calibration (lower ECE / NLL / Brier) give smaller conformal prediction sets?

Preliminary evidence from the report: the deep ensemble lowers NLL by 26% and ECE by 47% vs. the baseline, yet the set size is unchanged (1.20 vs 1.19 classes at 90% coverage). Here we test whether that holds across **all methods and 5 seeds**.

**Design**
- Every method is trained on **clean CIFAR-10** (`noise_ratio=0`), seeds **0-4**, with `train_with_noise` from `pipeline`, then evaluated with `evaluate_conformal`. This is the same path as `noise_testing.ipynb`, and it uses the same model configs, so existing clean (`noise_ratio=0`) checkpoints and result files in `models/noise_testing/` and `results/noise_testing/` are reused.
- **Calibration metrics** (accuracy, NLL, ECE, Brier) and **set size at α = 0.1** (randomized APS and LAC) come straight from the result dicts `evaluate_conformal` returns (the same numbers it writes to `results/noise_testing/{tag}.json`).
- **Threshold-free efficiency:** we call `evaluate_conformal` on each trained model at every α in a grid and take the **area under the set-size-vs-α curve**. The α = 0.1 call is the "official" one (it saves `{tag}.json` / `{tag}.npz`, like the noise notebook); the other α calls use `save=False`.
- **Rank correlations** (Spearman, Kendall) between every calibration metric and every efficiency metric, at four levels (see Section 3).

Run from the **repo root**. Training uses CUDA when available. The Laplace post-hoc fit moves to CPU by itself, which is fine.

```python
# Setup: only the pipeline package (+ common for EPOCHS / DEVICE)
import json
import numpy as np
import torch
import matplotlib.pyplot as plt
from scipy import stats
from IPython.display import Markdown, display

from pipeline import (Baseline, MCDropout, LaplaceLL, SNGP, DeepEnsemble, REPO,
                      train_with_noise, evaluate_conformal, compare, summarize)
from pipeline.evaluate import to_markdown
import common

print("device:", common.DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("WARNING: CUDA not available - training will fall back to", common.DEVICE,
          "and be MUCH slower. In Colab: Runtime > Change runtime type > GPU.")
```

```python
# ------------------------------- Config ---------------------------------------
SMOKE = False           # True: tiny subsets + 1 epoch to check everything runs end-to-end
EPOCHS = 1 if SMOKE else common.EPOCHS          # 40 = the shared recipe
SEEDS = [0, 1, 2, 3, 4]
NOISE_RATIOS = [0.0]    # clean CIFAR-10 only; the noise study is Problem 1
ALPHA = 0.1             # target coverage 90% (the alpha evaluate_conformal is run at)

# Grid for the set-size-vs-alpha curve (must contain ALPHA). AUC is normalised by the alpha
# range, so it reads as "mean set size over alpha in [0.02, 0.30]". Each extra alpha costs one
# more full inference pass per run (evaluate_conformal re-predicts), so keep the grid modest;
# MC Dropout (T=50) and Laplace (CPU) are the slow ones.
ALPHAS = np.round(np.linspace(0.02, 0.30, 15), 4)
assert ALPHA in ALPHAS
SCORES = ("aps_rand", "lac")

OUT_DIR = REPO / "results" / "calibration_efficiency"   # this notebook's own outputs
OUT_DIR.mkdir(parents=True, exist_ok=True)

# SAME configs as noise_testing.ipynb so existing clean checkpoints are reused
# (train_with_noise raises if a checkpoint's model config differs).
MODELS = {
    "baseline":      Baseline,
    "mc_dropout":    lambda: MCDropout(p=0.5, T=50),
    "laplace":       LaplaceLL,
    "sngp":          lambda: SNGP(norm_bound=6.0),
    "deep_ensemble": lambda: DeepEnsemble(n=5),
}
RUN_MODELS = ["baseline", "mc_dropout", "laplace", "sngp", "deep_ensemble"]

CALIB_KEYS = ["ece", "nll", "brier"]
EFF_KEYS = ["aps_set_size", "aps_auc", "lac_set_size", "lac_auc"]   # *_set_size is at ALPHA
ALL_KEYS = ["accuracy"] + CALIB_KEYS + EFF_KEYS
```

## 1. Train + evaluate every (method, seed), at every α

For each (method, seed): `train_with_noise` trains the model (or loads its checkpoint), then `evaluate_conformal` is called at α = 0.1 (saves `{tag}.json` / `{tag}.npz` to `results/noise_testing/`, exactly like the noise notebook) and again at every other α in `ALPHAS` with `save=False` to build the set-size-vs-α curve. The seed is the same for every α, so MC Dropout / SNGP sampling and the randomized-APS draws are identical across the curve.

**Cost:** training is 5 methods × 5 seeds, and the ensemble trains 5 networks per run, so 45 ResNet-20s at 40 epochs each (checkpoints are reused). The α sweep adds `len(ALPHAS)` inference passes per run. Try `SMOKE = True` first.

```python
results = []     # evaluate_conformal result at ALPHA, one per (method, seed)
curves = []      # {"aps_rand": {"sizes": [...], "coverage": [...]}, "lac": ...} per run, aligned with results
i_a = int(np.argmin(np.abs(ALPHAS - ALPHA)))

for name in RUN_MODELS:
    print(f"==================== {name}")
    for s in SEEDS:
        run = train_with_noise(MODELS[name](), noise_ratio=NOISE_RATIOS[0], seed=s,
                               epochs=EPOCHS, smoke=SMOKE)
        model, L = run["model"], run["loaders"]

        # Send the model through the conformal layer
        res = evaluate_conformal(model, L["cal"], L["test"], alpha=ALPHA, seed=s, n_resplits=0,
                                 tag=run["tag"], config=run["config"])
        results.append(res)

        # set-size-vs-alpha curve: same function, other alphas, nothing saved
        cur = {sc: {"sizes": [], "coverage": []} for sc in SCORES}
        for j, a in enumerate(ALPHAS):
            r = res if j == i_a else evaluate_conformal(model, L["cal"], L["test"], alpha=float(a),
                                                        seed=s, n_resplits=0, tag=run["tag"],
                                                        config=run["config"], save=False)
            for sc in SCORES:
                cur[sc]["sizes"].append(r["conformal"][sc]["avg_set_size"])
                cur[sc]["coverage"].append(r["conformal"][sc]["coverage"])
        curves.append(cur)
        print(f"[{run['tag']}] done: APS size@{ALPHA}={res['conformal']['aps_rand']['avg_set_size']:.3f}")

        del run, model, L
        if torch.cuda.is_available():
            torch.cuda.empty_cache()        # release GPU memory (ensembles hold 5 nets)
print(f"\n{len(results)} runs")
```

```python
# One flat row per run. NOTE: path is set so this does NOT overwrite the noise notebook's
# results/noise_testing/results.json (compare() writes there by default).
rows = compare(results, path=OUT_DIR / "runs_full_results.json")
display(Markdown(to_markdown(rows, ["model", "seed", "accuracy", "nll", "ece", "brier",
                                    "aps_coverage", "aps_set_size", "lac_coverage", "lac_set_size"])))
```

## 2. Threshold-free efficiency: area under the set-size-vs-α curve

The curves from Section 1 are integrated (trapezoid rule) and divided by the α range, giving the **mean set size over the α grid** for each run.

```python
_trapz = getattr(np, "trapezoid", None) or np.trapz     # np.trapz was renamed in NumPy 2

for row, cur in zip(rows, curves):
    row["aps_auc"] = float(_trapz(cur["aps_rand"]["sizes"], ALPHAS) / (ALPHAS[-1] - ALPHAS[0]))
    row["lac_auc"] = float(_trapz(cur["lac"]["sizes"], ALPHAS) / (ALPHAS[-1] - ALPHAS[0]))
print("AUC added to", len(rows), "runs")
```

```python
# Clean vs. method summary: mean ± std over seeds (summarize() from evaluate.py)
summary = summarize(rows)
display(Markdown(to_markdown(summary, ["model", "n_seeds", "accuracy", "nll", "ece", "brier",
                                       "aps_coverage", "aps_set_size", "lac_coverage", "lac_set_size"])))

# AUC summary in the same format
auc_summary = []
for m in RUN_MODELS:
    g = [r for r in rows if r["model"] == m]
    d = {"model": m, "n_seeds": len(g)}
    for k in ("aps_auc", "lac_auc"):
        v = np.array([r[k] for r in g]); d[k] = float(v.mean()); d[k + "_std"] = float(v.std(ddof=1)) if len(v) > 1 else 0.0
    auc_summary.append(d)
display(Markdown(to_markdown(auc_summary)))
```

## 3. Why rank correlations?

The claim from the report is about **orderings**: ranking methods by ECE/NLL may not predict their ranking by conformal efficiency. A rank correlation tests exactly that.

- **Ordinal question.** Spearman ρ and Kendall τ ask "when one run is better calibrated than another, is it also more efficient?". Only the order matters, so it needs no assumption that the relationship is linear, and it is unaffected by the very different scales (ECE ≈ 0.02, set size ≈ 1.2).
- **Robust with few points.** With 25 runs (5 methods × 5 seeds), single outliers (e.g. a badly calibrated seed) would dominate a Pearson correlation or a regression slope. Ranks are much less sensitive to that.
- **A table of means can't answer it.** Comparing two methods (ensemble vs. baseline) gives one data point. Correlating across all methods *and* seeds gives a quantified, repeatable answer, with a CI, and shows whether seed-to-seed variation within a method points the same way.
- **Calibration and efficiency measure different things.** Calibration is about whether probabilities match frequencies. Set size depends mostly on how well the model *ranks* the true class near the top (accuracy), and conformal coverage is guaranteed regardless of calibration. So a weak correlation would be a meaningful finding, not a failure.

**How to read it.** ρ near 0 means calibration ranking does not predict efficiency ranking. A positive ρ (worse calibration ↔ larger sets) means it does. Correlation is not causation, and accuracy is a confounder for both sides, so we report four levels:
- **pooled:** all (method, seed) runs together, n = 25. Dominated by between-method differences.
- **within_method:** each metric de-meaned per method, then pooled. Do better-calibrated *seeds of the same method* get smaller sets?
- **method_means:** the five per-method means (low power, descriptive).
- **partial_acc:** pooled Spearman after regressing ranks on accuracy, i.e. what calibration tells us about set size *beyond* accuracy.

```python
def col(rs, k): return np.array([r[k] for r in rs], dtype=float)

def spearman(x, y):
    if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0: return float("nan"), float("nan")
    r, p = stats.spearmanr(x, y); return float(r), float(p)

def kendall(x, y):
    if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0: return float("nan"), float("nan")
    r, p = stats.kendalltau(x, y); return float(r), float(p)

def partial_spearman(x, y, z):
    # Spearman of x, y after linearly removing the rank of z from each rank.
    rx, ry, rz = (stats.rankdata(v) for v in (x, y, z))
    A = np.column_stack([np.ones_like(rz), rz])
    res = lambda r: r - A @ np.linalg.lstsq(A, r, rcond=None)[0]
    ex, ey = res(rx), res(ry)
    if np.std(ex) == 0 or np.std(ey) == 0: return float("nan")
    return float(np.corrcoef(ex, ey)[0, 1])

def boot_ci(x, y, n_boot=2000, seed=0):
    # Bootstrap 95% CI of Spearman, resampling runs.
    rng, rs, n = np.random.RandomState(seed), [], len(x)
    for _ in range(n_boot):
        i = rng.randint(0, n, n)
        if np.std(x[i]) > 0 and np.std(y[i]) > 0:
            rs.append(stats.spearmanr(x[i], y[i])[0])
    return [float(np.percentile(rs, 2.5)), float(np.percentile(rs, 97.5))] if rs else [np.nan, np.nan]


def correlations(rows):
    methods = list(dict.fromkeys(r["model"] for r in rows))
    out = {"pooled": {}, "within_method": {}, "method_means": {}, "partial_acc": {}}

    dem = {k: np.zeros(len(rows)) for k in ALL_KEYS}          # per-method de-meaned metrics
    for m in methods:
        idx = [i for i, r in enumerate(rows) if r["model"] == m]
        for k in ALL_KEYS:
            v = col(rows, k); dem[k][idx] = v[idx] - v[idx].mean()
    mean_rows = [{"model": m, **{k: float(np.mean([r[k] for r in rows if r["model"] == m]))
                                 for k in ALL_KEYS}} for m in methods]

    for c in CALIB_KEYS:
        for e in EFF_KEYS:
            key = f"{c}|{e}"
            x, y = col(rows, c), col(rows, e)
            rs, ps = spearman(x, y); rk, pk = kendall(x, y)
            out["pooled"][key] = {"n": len(x), "spearman": rs, "spearman_p": ps,
                                  "spearman_ci95": boot_ci(x, y), "kendall": rk, "kendall_p": pk}
            rs, ps = spearman(dem[c], dem[e])
            out["within_method"][key] = {"n": len(x), "spearman": rs, "spearman_p": ps}
            rs, ps = spearman(col(mean_rows, c), col(mean_rows, e))
            out["method_means"][key] = {"n": len(mean_rows), "spearman": rs, "spearman_p": ps}
            out["partial_acc"][key] = {"n": len(x), "spearman": partial_spearman(x, y, col(rows, "accuracy"))}
    out["accuracy_vs_eff"] = {e: dict(zip(("spearman", "spearman_p"), spearman(col(rows, "accuracy"), col(rows, e))))
                              for e in EFF_KEYS}
    return out, mean_rows

corr, mean_rows = correlations(rows)
```

```python
def md_table(headers, body):
    lines = ["| " + " | ".join(headers) + " |", "|" + "---|" * len(headers)]
    return "\n".join(lines + ["| " + " | ".join(r) + " |" for r in body])

LEVELS = [("pooled", "Pooled over all (method, seed) runs"),
          ("within_method", "Within-method (method means removed)"),
          ("method_means", "Across method means (low power)"),
          ("partial_acc", "Pooled, partialling out accuracy")]

for level, title in LEVELS:
    body = []
    for c in CALIB_KEYS:
        row = [c]
        for e in EFF_KEYS:
            d = corr[level][f"{c}|{e}"]
            s = f"{d['spearman']:+.2f}"
            if "spearman_p" in d and not np.isnan(d["spearman_p"]): s += f" (p={d['spearman_p']:.3f})"
            row.append(s)
        body.append(row)
    n = next(iter(corr[level].values()))["n"]
    display(Markdown(f"**Spearman ρ - {title} (n={n})**\n\n" +
                     md_table(["calibration \\ efficiency"] + EFF_KEYS, body)))

body = [[c] + [f"[{corr['pooled'][f'{c}|{e}']['spearman_ci95'][0]:+.2f}, {corr['pooled'][f'{c}|{e}']['spearman_ci95'][1]:+.2f}]"
               for e in EFF_KEYS] for c in CALIB_KEYS]
display(Markdown("**Pooled Spearman, bootstrap 95% CI over runs**\n\n" +
                 md_table(["calibration \\ efficiency"] + EFF_KEYS, body)))

body = [[c] + [f"{corr['pooled'][f'{c}|{e}']['kendall']:+.2f} (p={corr['pooled'][f'{c}|{e}']['kendall_p']:.3f})"
               for e in EFF_KEYS] for c in CALIB_KEYS]
display(Markdown("**Pooled Kendall τ**\n\n" + md_table(["calibration \\ efficiency"] + EFF_KEYS, body)))

display(Markdown("**Accuracy vs efficiency (confounder check)**\n\n" + md_table(
    ["efficiency metric", "Spearman ρ", "p"],
    [[e, f"{d['spearman']:+.2f}", f"{d['spearman_p']:.3f}"] for e, d in corr["accuracy_vs_eff"].items()])))
```

## 4. Plots

```python
# Set size vs alpha, mean +- std over seeds, per method
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, score, title in zip(axes, SCORES, ["Randomized APS", "LAC"]):
    for m in RUN_MODELS:
        S = np.array([cur[score]["sizes"] for cur, r in zip(curves, rows) if r["model"] == m])
        ax.plot(ALPHAS, S.mean(0), label=m)
        ax.fill_between(ALPHAS, S.mean(0) - S.std(0), S.mean(0) + S.std(0), alpha=0.2)
    ax.axvline(ALPHA, ls="--", c="gray", lw=0.8)
    ax.set_title(f"{title}: set size vs alpha"); ax.set_xlabel("alpha"); ax.set_ylabel("avg set size")
axes[0].legend(); plt.tight_layout(); plt.show()
```

```python
# Calibration metric vs efficiency (AUC), each point = one (method, seed) run
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for r_i, (eff, elab) in enumerate([("aps_auc", "APS set-size AUC"), ("lac_auc", "LAC set-size AUC")]):
    for c_i, c in enumerate(CALIB_KEYS):
        ax = axes[r_i, c_i]
        for m in RUN_MODELS:
            g = [r for r in rows if r["model"] == m]
            ax.scatter(col(g, c), col(g, eff), label=m, s=40)
        rho = corr["pooled"][f"{c}|{eff}"]["spearman"]
        ax.set_title(f"{c.upper()} vs {elab}  (pooled ρ={rho:+.2f})")
        ax.set_xlabel(c); ax.set_ylabel(elab)
axes[0, 0].legend(); plt.tight_layout(); plt.show()
```

```python
# Heatmaps of Spearman rho: pooled vs partial (accuracy removed)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.6))
for ax, level, title in zip(axes, ["pooled", "partial_acc"], ["Pooled", "Accuracy partialled out"]):
    M = np.array([[corr[level][f"{c}|{e}"]["spearman"] for e in EFF_KEYS] for c in CALIB_KEYS])
    im = ax.imshow(M, vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(EFF_KEYS))); ax.set_xticklabels(EFF_KEYS, rotation=30, ha="right")
    ax.set_yticks(range(len(CALIB_KEYS))); ax.set_yticklabels(CALIB_KEYS)
    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            ax.text(j, i, f"{M[i, j]:+.2f}", ha="center", va="center", fontsize=9)
    ax.set_title(f"Spearman ρ - {title}")
fig.colorbar(im, ax=axes, shrink=0.8); plt.show()
```

```python
# Report-style comparison: ensemble vs baseline (the Table VII claim), mean over seeds
b = {k: col([r for r in rows if r["model"] == "baseline"], k).mean() for k in ALL_KEYS}
e = {k: col([r for r in rows if r["model"] == "deep_ensemble"], k).mean() for k in ALL_KEYS}
for k in CALIB_KEYS:
    print(f"{k:6s}: ensemble vs baseline {100 * (e[k] - b[k]) / b[k]:+.1f}%")
for k in ["aps_set_size", "lac_set_size", "aps_auc", "lac_auc"]:
    print(f"{k:13s}: baseline {b[k]:.3f}  ensemble {e[k]:.3f}  ({100 * (e[k] - b[k]) / b[k]:+.1f}%)")
```

## Notes
- Split-conformal coverage is about 1 - α whatever the model, so differences between methods appear in **set size**, not coverage. `aps_coverage` / `lac_coverage` in the tables should be ≈0.90; a large deviation means a bug.
- With n = 25 runs that cluster by method, pooled p-values are optimistic. Lean on the bootstrap CIs, the within-method correlations and the accuracy-partialled correlations. The method-means correlation has only n = 5 points.
- The AUC is the mean set size over α ∈ [0.02, 0.30]; changing `ALPHAS` only re-runs inference (checkpoints are reused), but each extra α is another full `evaluate_conformal` pass per run.
- This notebook shares `results/noise_testing/{tag}.json|npz` with `noise_testing.ipynb` for the clean runs (same tag, e.g. `baseline_noise0_seed0`). If the noise notebook was run with `N_RESPLITS = 100`, re-running here at `n_resplits=0` rewrites those files without the resplit fields.
- For exact run-to-run GPU determinism, pass `deterministic=True` to `train_with_noise`.
