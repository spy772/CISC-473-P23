# Import SVHN (out-of-distribution) dataset

```python
# Define basic normalization for SVHN (typically mean/std of digits)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4377, 0.4438, 0.4728), (0.1980, 0.2010, 0.1970))
])

# Download the training split
train_set = torchvision.datasets.SVHN(
    root='./data_svhn',
    split='train',
    download=True,
    transform=transform
)

# Download the testing split
test_set = torchvision.datasets.SVHN(
    root='./data_svhn',
    split='test',
    download=True,
    transform=transform
)

print(f"SVHN Train Size: {len(train_set)} images") # 73,257 digits
print(f"SVHN Test Size: {len(test_set)} images")   # 26,032 digits
```

```python
import scipy.io as sio
import numpy as np
import os

# Load the raw Stanford SVHN file
mat_path = './data_svhn/train_32x32.mat'
data = sio.loadmat(mat_path)

X = data['X']  # Shape: (32, 32, 3, 73257)
y = data['y']  # Shape: (73257, 1)

# Split into 4 chunks (~18,314 samples per chunk)
num_chunks = 4
X_splits = np.array_split(X, num_chunks, axis=3)
y_splits = np.array_split(y, num_chunks, axis=0)

# Save each chunk as a separate small .mat file
os.makedirs('./data_svhn_chunks', exist_ok=True)
for i in range(num_chunks):
    chunk_data = {
        'X': X_splits[i],
        'y': y_splits[i]
    }
    sio.savemat(f'./data_svhn_chunks/train_chunk_{i}.mat', chunk_data)

print("Successfully chunked SVHN! Each file is safely under 50 MB.")
```
