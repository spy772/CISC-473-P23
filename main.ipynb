```python
import torch

import torchvision
import torchvision.transforms as transforms
```

# Import dataset

```python
# Define transforms if needed
transform = transforms.Compose([transforms.ToTensor()])

# This downloads CIFAR-10 to a './data' folder automatically
train_dataset = torchvision.datasets.CIFAR10(
    root='./data',
    train=True,
    download=True,
    transform=transform
)
```

# Import ResNet-20 model

```python
# Load the exact paper-spec ResNet-20 model locally
model = torch.hub.load('akamaster/pytorch_resnet_cifar10', 'resnet20', pretrained=False)

# Ready for local training loop
print(f"Total parameters: {sum(p.numel() for p in model.parameters())}") # Should be ~270,000
```
